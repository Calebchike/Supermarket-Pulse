# **Import Libraries**

```python
# Import necessary libraries: numpy for numerical operations and pandas for data manipulation.
import numpy as np
import pandas as pd
```

# **Load the CSV data**
Load Raw dataset, determine the rows and columns and convert columns to Panda series

```python
# Load the supermarket sales data from a CSV file into a pandas DataFrame.
# The 'Invoice ID' column is set as the index.
# The .head() method is used to display the first few rows of the DataFrame.
supermarket_raw = pd.read_csv("https://raw.githubusercontent.com/Calebchike/Supermarket-Pulse/refs/heads/main/SuperMarket%20Analysis.csv", index_col="Invoice ID") #import supermarket csv file as a dataframe
supermarket_raw.head()                                         #prints the dataframe
```

### Get the Number of Rows and Columns contained in the dataset

```python
# Display the number of rows and columns in the DataFrame using the .shape attribute.
supermarket_raw.shape     #get the number of rows and column in the data
```

### Get all the column Headers in a List for referencing

```python
# Get a list of all column headers (names) from the DataFrame.
supermarket_raw.columns.to_list()                         #prints all the heading of the dataframe as a list
```

This cell creates Panda series for the Columns

```python
# Create individual pandas Series for each column in the raw supermarket data.
# Each Series is indexed by 'Invoice ID' and named after its respective column.
branch = pd.Series(supermarket_raw["Branch"], index = supermarket_raw.index, name="Branch")
city_custtype = pd.Series(supermarket_raw["City_CustType"], index=supermarket_raw.index, name="City_CustType")
gender = pd.Series(supermarket_raw["Gender"], index=supermarket_raw.index, name="Gender")
product_line = pd.Series(supermarket_raw["Product line"], index=supermarket_raw.index, name="Product line")
unit_price = pd.Series(supermarket_raw["Unit price"], index=supermarket_raw.index, name="Unit price")
quantity = pd.Series(supermarket_raw["Quantity"], index=supermarket_raw.index, name="Quantity")
tax = pd.Series(supermarket_raw["Tax 5%"], index=supermarket_raw.index, name="Tax")
sales = pd.Series(supermarket_raw["Sales"], index=supermarket_raw.index, name="Sales")
date = pd.Series(supermarket_raw["Date"], index=supermarket_raw.index, name="Date")
time = pd.Series(supermarket_raw["Time"], index=supermarket_raw.index, name="Time")
payment = pd.Series(supermarket_raw["Payment"], index=supermarket_raw.index, name="Payment")
cogs = pd.Series(supermarket_raw["cogs"], index=supermarket_raw.index, name="cogs")
gross_margin_percentage = pd.Series(supermarket_raw["gross margin percentage"], index=supermarket_raw.index, name="gross margin percentage")
gross_income = pd.Series(supermarket_raw["gross income"], index=supermarket_raw.index, name="gross income")
rating = pd.Series(supermarket_raw["Rating"], index=supermarket_raw.index, name="Rating")
```

# **Data Cleaning**

## Split Column
This cell Split "City_CustType" column into two separate columns using the |
delimiter. and Named them "City" and "Customer Type".

```python
# Split the 'City_CustType' Series into two new Series: 'City' and 'Customer Type',
# using the '|' delimiter. Then display the first few entries of each new Series.
city = city_custtype.str.split("|").str[0]
customer_type = city_custtype.str.split("|").str[1]
city.name = "City"
customer_type.name = "Customer Type"
display(city.head())
display(customer_type.head())
```

## Standardize Text
The Product line column has inconsistent capitalization. This cells Forces all text in the column to "Capitalize Each Word".

```python
# Display the first few entries of the 'product_line' Series to show its current capitalization.
product_line.head()
```

```python
# Standardize the 'product_line' Series by stripping any leading/trailing whitespace
# and then capitalizing the first letter of each word in the strings.
# Finally, display the first few entries of the standardized Series.
product_line = product_line.str.strip().str.title()
product_line.head()
```

## Handling Nulls
 This cell Filters out all transactions where the customer Rating was left blank.

```python
# Display the first few entries of the 'rating' Series and its total size.
rating.head()
rating.size
```

```python
# Identify and store the indices of transactions where 'rating' is null.
# Then, drop these null-rating transactions from all relevant Series to maintain data consistency.
# Finally, display the first few entries and the new size of the 'rating' Series.
null_rating = [x for x in rating.index if pd.isnull(rating[x])]

branch = branch.drop(null_rating)
city = city.drop(null_rating)
customer_type = customer_type.drop(null_rating)
gender = gender.drop(null_rating)
product_line = product_line.drop(null_rating)
unit_price = unit_price.drop(null_rating)
quantity = quantity.drop(null_rating)
tax = tax.drop(null_rating)
sales = sales.drop(null_rating)
date = date.drop(null_rating)
time = time.drop(null_rating)
payment = payment.drop(null_rating)
cogs = cogs.drop(null_rating)
gross_margin_percentage = gross_margin_percentage.drop(null_rating)
gross_income = gross_income.drop(null_rating)
rating = rating.drop(null_rating)

rating.head()
rating.size
```

# **Baseline Metrics & Margin of Error**

## Measures of Central Tendency:
Calculate the Mean, Median, and Mode for the Sales transaction
value.

```python
# Calculate and print the Mean, Median, and Mode of the 'sales' Series.
# The mode might return multiple values, so it's converted to a list for display.
print(f"The Mean of the Total Sales is: {sales.mean()}")
print(f"The Median of the Total Sales is: {sales.median()}")
print(f"The Mode of the Total Sales are: {sales.mode().to_list()}")
```

## Measures of Dispersion:
Calculate the sample Variance and sample Standard Deviation for the customer Rating column.

```python
# Calculate and print the sample Variance and Standard Deviation of the 'rating' Series,
# formatted to 4 significant figures.
print(f"The Variance of the Rating is: {rating.var():.4g}")
print(f"The Standard Deviation of the Rating is: {rating.std():.4g}")
```

## Margin of Error:
Take a random sample of 60 transactions. Calculate the Standard Error and Margin of Error for the Total sales value, assuming a 95% confidence level (Z-score = 1.96).

```python
# Take a random sample of 60 transactions from the 'sales' Series.
# Calculate and print the Mean, Standard Error (sem), and Margin of Error (assuming a 95% confidence level with Z-score = 1.96)
# for this random sample, formatted to 4 decimal places.
random_sixty = sales.sample(60)
print(f"The Mean of the Random Sample is: {random_sixty.mean():.4f}")
print(f"The Standard Error of the Random Sample is: {random_sixty.sem():.4f}")
print(f"The Margin of Error of the Random Sample is: {1.96 * random_sixty.sem():.4f}")
```

# **Probability Distributions**

## The Normal Distribution (Sales Revenue)

What is the probability that the next customer's total bill will be $500 or less?

```python
# Import pandas for data manipulation and scipy.stats.norm for normal distribution calculations.
import pandas as pd
from scipy.stats import norm

# 1. Calculate the mean and standard deviation from your Sales column.
mean_sales = sales.mean()
std_sales = sales.std()

# 2. Calculate the cumulative probability (CDF) that a random value from a normal distribution
# with the calculated mean and standard deviation will be 500 or less.
probability = norm.cdf(500, loc=mean_sales, scale=std_sales)

# 3. Print the result as a percentage, formatted to two decimal places.
print(f"There is a {probability * 100:.2f}% chance the next bill will be $500 or less.")
```
